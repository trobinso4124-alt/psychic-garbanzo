```python
import yt_dlp
from IPython.display import Video, display, HTML
import ipywidgets as widgets
```

```python
# YouTube video downloader and player
def download_and_play(url):
    ydl_opts = {
        'format': 'best[height<=720]',
        'quiet': True,
        'no_warnings': True
    }
    
    try:
        with yt_dlp.YoutubeDL(ydl_opts) as ydl:
            info = ydl.extract_info(url, download=False)
            video_url = info['url']
            title = info.get('title', 'Video')
            print(f"Now playing: {title}")
            display(Video(video_url, width=640, height=480))
    except Exception as e:
        print(f"Error: {e}")

# Interactive widget
url_input = widgets.Text(
    value='https://www.youtube.com/watch?v=dQw4w9WgXcQ',
    placeholder='Enter YouTube URL',
    description='URL:',
    layout=widgets.Layout(width='80%')
)

play_button = widgets.Button(
    description='Play Video',
    button_style='success'
)

output = widgets.Output()

def on_button_click(b):
    with output:
        output.clear_output()
        download_and_play(url_input.value)

play_button.on_click(on_button_click)

display(HTML("<h3>YouTube Video Player using yt-dlp</h3>"))
display(url_input, play_button, output)
```
